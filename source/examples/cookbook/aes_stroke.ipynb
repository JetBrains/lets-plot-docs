# `stroke` aesthetic

Affects the thickness of the point boundaries (in case the given shape has a boundary).

Available for the following geometries: `geom_point()`, `geom_jitter()`, `geom_qq()`, `geom_qq2()`, `geom_pointrange()`, `geom_dotplot()`, `geom_ydotplot()`.

```python
from lets_plot import *
```

```python
LetsPlot.setup_html()
```

```python
data = {
    'x': (list(range(7)) * 4)[:26],
    'y': ([3] * 7 + [2] * 7 + [1] * 7 + [0] * 7)[:26],
    'shape': list(range(26)),
}
```

```python
p = ggplot(data, aes('x', 'y')) + scale_shape_identity() + \
    xlim(-1, 7) + ylim(-1, 4) + \
    theme(legend_position='none')
```

#### 1. Default Presentation

```python
p + geom_point(aes(shape='shape'), size=12, color="#54278f", fill="#dd1c77")
```

#### 2. `stroke=8`

```python
p + geom_point(aes(shape='shape'), size=12, stroke=8, color="#54278f", fill="#dd1c77")
```
