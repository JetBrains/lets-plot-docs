# Discrete Scales

Map discrete values to visual values.

```python
import pandas as pd

from lets_plot import *
```

```python
LetsPlot.setup_html()
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/JetBrains/lets-plot-docs/master/data/mpg.csv')
```

```python
p = ggplot(df, aes(x='cyl')) + geom_bar(aes(fill='cyl'))
p1 = p + ggtitle('Default')
p2 = p + scale_fill_discrete() + ggtitle('With Scale')

gggrid([p1, p2])
```
