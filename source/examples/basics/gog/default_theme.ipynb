# Default Theme

See [theme()](https://jetbrains.github.io/lets-plot-docs/pages/api/lets_plot.theme.html#lets_plot.theme).

```python
import pandas as pd

from lets_plot import *
LetsPlot.setup_html()
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/JetBrains/lets-plot-docs/master/data/mpg.csv')
```

```python
p = ggplot(df, aes(x='fl')) + geom_bar()

w, h = 400, 300
bunch = GGBunch()
bunch.add_plot(p + ggtitle("Default Theme"), 0, 0, w, h)
bunch.add_plot(p + ggtitle("Customized Theme") + \
               theme(axis_line_x='blank', axis_ticks_x='blank', \
                     panel_grid_major_x='blank', panel_grid_major_y=element_line(size=2)), w, 0, w, h)
bunch.show()
```
